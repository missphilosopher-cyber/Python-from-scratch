```python
import pandas as pd
```

```python
marks=pd.Series([95,87,92,76,89])
print(marks)
```

```python
#Assessing elements in a Series
print(marks[2])
```

```python
#Custom indexing in Pandas Series
import pandas as pd

marks=pd.Series([95,87,92],index=["Aayu","Riya","Arjun"])

print(marks)
```

```python
#Accessing values using labels
print(marks["Riya"])
```

```python
#Understanding .loc and .iloc
#.loc means label-based-selection and .iloc means position-based-selection
a=marks.loc["Riya"]
print(a)
b=marks.iloc[1]
print(b)
```

```python
#DATAFRAMES:
#Dataframes: Two dimensional lablled data, arranged in rows and columns(like an excel table inside python)
#Series are only one dimensional data hence differ from dataframes
```

```python
import pandas as pd
data={
    "Name":["Aayu","Riya","Arjun","Kabir"],
    "Age": [18,19,18,20],
    "Marks":[95,87,92,76]
}

df=pd.DataFrame(data)

df
```

```python
#NOTE: While construction a Dataframe from dictionaries of lists, all the lists must have the same length.
```

```python
#ACCESSING A SINGLE COLUMN IN A DATAFRAME
print(df["Marks"])
```

```python
print(type(df))
print(type(df["Marks"]))
```

```python
#SELECTING ROWS AND COLUMNS
import pandas as pd
data={
    "Name":["Aayu","Riya","Arjun","Kabir"],
    "Age": [18,19,18,20],
    "Marks":[95,87,92,76]
}

df=pd.DataFrame(data)

df
```

```python
df[["Name","Marks"]]
```

```python
df.iloc[0]
#aince the row mixes strings and integers, the resulting series has object datatype
```

```python
df.loc[2]
```

```python
df.index=[10,20,30,40]
```

```python
df
```

```python
#selecting a particular row & column
df.loc[20,"Marks"]
```

```python
print(df.iloc[2,2])
```

```python
#SLICING ROWS AND COLUMNS IN PANDAS
#Slicing means selecting a portion of data instead of selecting just one element
import pandas as pd
data={
    "Name":["Aayu","Riya","Arjun","Kabir","Meera"],
    "Age": [18,19,18,20,19],
    "Marks":[95,87,92,76,89]
}

df=pd.DataFrame(data)

df
```

```python
df.iloc[1:4]
```

```python
df.loc[1:3] #.loc slices include both endpoints
```
